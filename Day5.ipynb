**Dict.** **Method**

```python
#.Keys()
d1 ={'Name':'Dev',"Age":20}
print(d1)
print(d1.keys())
```

```python
#.values()
d2={'Name':'Devendra','Age':20}
print(d2)
print(d2.values())
```

**Loops**

```python
#for
for i in range(6):
  print(i)
```

```python
l4 = [20,30,40,60]
for i in l4:
  print(i)
```

```python
d4={'Name':'Dev',"Age":20}
for i in d4:
  print(i)

  print(d4[i])

for i in d4.items():
  print(i)

for i in d4.keys():
  print(d4[i])
```

```python
d6 = {'user':['nikhil','raj']}
for i in range(len(d6['user'])):
  print(d6['user'][i])


for i in d6["user"]:
    print(i)
```

```python
#while loop
i = 1

while i <= 5:
    print(i)
    i += 1
```

**Fucntion**

```python
def hello():
  print('Hello_Function')
hello()
```

```python
def Name(a):
  print(a)

Name('Devendra')
```

```python
def Name(a,b):
  print(a,b)
Name('Devendra','Kumawat')
```

```python
def name(*a):
  print(a)
name('Devendra')
name('Devendra','Kumawat')
```

```python
def name(*a):
  for i in a:
    print(i)

name('Devendra')
name('Devendra','Kumawat')
p = ['raj',20,30]
name(*p)
```

```python
def atm(a):
  return a

b = atm(200)
print(b)
```
