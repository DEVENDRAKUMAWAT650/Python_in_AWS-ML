```python
#Rshape
#1d arry
import numpy as np
arr = np.arange(6)
print(arr)

#2D
arr = arr.reshape(2,3)
print(arr)


arr1 = np.arange(15).reshape(5,3)
print(arr1)

#3D
arr2 = np.arange(24).reshape(2,3,4)
print(arr2)
```

```python
#flatten  || make a copy in memory
arrnew = arr.flatten()
arr = np.arange(6).reshape(2,3)
print(arr)
arr = arr.flatten()
print(arr)
arrnew = arr.flatten()
arrnew[0]=100
print(arr)
```

```python
#ravel || original me change
arr = np.arange(6).reshape(2,3)
print(arr)
arr = arr.ravel()
print(arr)
arrnew = arr.ravel()
print(arrnew)
arrnew[0]=100
print(arr)
```

```python
#transpose
arr = np.arange(6).reshape(2,3)
print(arr)
arr = arr.transpose()
print(arr)
arr1 = arr.T
print(arr1)
```
