```python
import pandas as pd

data = {
    "Emp_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Riya", "Raj", "Sara", "John", "Neha"],
    "Department": ["IT", "HR", "Finance", "IT", "Sales", "HR"],
    "Salary": [50000, 45000, 60000, 55000, 48000, 52000],
    "Experience": [2, 3, 5, 4, 1, 3]
}

df = pd.DataFrame(data)
print(df)

# loc for indexing(row ka index,second->cols name)
print(df.loc[1,'Name'])
```

```python
#iloc ->lables(row,cols)
print(df.iloc[1,1])
```

```python
data = {
    "Emp_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Riya", "Raj", "Sara", "John", "Neha"],
    "Department": ["IT", "HR", "Finance", "IT", "Sales", "HR"],
    "Salary": [50000, 45000, 60000, 55000, 48000, 52000],
    "Experience": [2, 3, 5, 4, 1, 3]
}

df = pd.DataFrame(data)
print(df)

# get rows
print(df.loc[[0,1,2]])
```

```python
#get rows using iloc
print(df.iloc[[0,1,2,3]])
```

```python
#mulit row and colum
print(df.iloc[[0,1],[0,1]])
```

```python
df.loc[0:2,"Name":"Salary"]
df.loc[0:2]
df.iloc[0:2]
```

```python
#filteration
data = {
    "Emp_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Riya", "Raj", "Sara", "John", "Neha"],
    "Department": ["IT", "HR", "Finance", "IT", "Sales", "HR"],
    "Salary": [50000, 45000, 60000, 55000, 48000, 52000],
    "Experience": [2, 3, 5, 4, 1, 3]
}

df = pd.DataFrame(data)
print(df)
```

```python
#garter then
print(df[df["Emp_ID"]>104])
```

```python
#for find out
print(df[df['Emp_ID']==106])
```

```python
print(df[(df['Salary']>30000) & (df["Salary"]<50000)])
```

```python
#sorting ascending
print(df.sort_values(by="Salary"))
# descending
print(df.sort_values(by="Salary",ascending=False))

#two cols
print(df.sort_values(['Salary','Experience']))
```
