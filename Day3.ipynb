**Operators**
1. Arithmatic +,-,*,%,**,//,/
2. Relational >,<,>=,<=,==,!=
3. Assignment +=,-+,*=,%=

5. Bitwise    &,|

**Arithmatic** **Operators**


```python
num1=10
num2=20
print(num1+num2)
print(num1-num2)
print(num1*num2)
print(num1/num2)
print(num1%num2)
print(num1**num2)
print(num1//num2)
```

**Relational** **Operator**

```python
a = 10
b = 20

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= 10)
print(b <= 20)
```

**Assignment** **Operator**

```python
x = 10

x += 5
print(x)

x -= 3
print(x)

x *= 2
print(x)

x /= 4
print(x)

x %= 2
print(x)

x **= 3
print(x)

x //= 2
print(x)
```

**Logical** **Operator**

```python
a = 10
b = 20
c = 30
print(not(a < b and b > c or c > a))
```

```python
#and
x = 10

print(x > 5 and x < 20)
print(x > 15 and x < 20)
```

```python
#or
x = 10

print(x > 5 or x > 20)
print(x < 5 or x > 20)
```

```python
#not
x = 10

print(not(x > 5))
print(not(x < 5))
```

**Bitwise** **Operator**

```python
#AND (&)
a = 5
b = 3

print(a & b)
```

```python
#OR (|)
a = 5      # 0101
b = 3      # 0011

print(a | b)
```
