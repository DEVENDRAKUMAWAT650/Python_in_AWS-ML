#Handle missing values

```python

data = {
    'exprinece': [ 1,2,3,4,5],
    'salary': [100,200,300,400,500]
}
df = pd.DataFrame(data)
df['exprinece'] = [1,np.nan,3,4,5]
df['salary'] = [100,200,300,np.nan,500]
df['name'] = ['hello','dev',"Nikhil",'raj','vinay']
df
```

```python

NULL_value = df.isnull()
print(NULL_value)
#sum of null
print(df.isnull().sum())

```

```python
#drop NULL value (drop complete row)
#df.dropna(inplace=True)
df.dropna()
df
```

```python
#fillna | null -> fill
df.fillna(5)
```

```python
import matplotlib.pyplot as plt

# Bar Chart Data
i1 = ['India', 'USA', 'UK']
v1 = [120, 90, 70]

# Ek hi figure
plt.figure(figsize=(14,7), dpi=150)

# ---------------- Line Chart ----------------
data = {
    "experience":[1,2,3,4,5],
    "salary":[100,200,300,400,500]
}

plt.subplot(2,2,1)
plt.plot(data["experience"], data["salary"], marker='o')
plt.title("Line Chart")

# ---------------- Pie Chart ----------------
fruits = ['apple','banana','carrot','mango']
values = [30,50,70,2]

plt.subplot(2,2,2)
plt.pie(values, labels=fruits, autopct="%1.1f%%")
plt.title("Pie Chart")

# ---------------- Bar Chart ----------------
plt.subplot(2,2,3)
plt.bar(i1, v1, color=['r','g','b'])
plt.xlabel("Country")
plt.ylabel("Count")
plt.title("Bar Chart")

# ---------------- Stacked Bar Chart ----------------
subjects = ['Math', 'Science', 'English', 'Computer']
boys = [80,75,90,85]
girls = [85,78,88,92]

plt.subplot(2,2,4)
plt.bar(subjects, boys, label='Boys')
plt.bar(subjects, girls, bottom=boys, label='Girls')

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Stacked Bar Chart")
plt.legend()

# Sabse last me
plt.tight_layout()
plt.show()
```
