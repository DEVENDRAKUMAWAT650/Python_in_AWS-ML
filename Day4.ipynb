**conditional** **Statement**

```python
#if
num1=10
if num1>5:
  print("num1 is greater than 5")
print("if End")
```

```python
num1 =20
num2 =10
if num1   <= num2:
  print("num1 is less than or quleto num2")
else:
  print("else part is Running")

print("if,else End")
```

```python
num1=int(input("Enter The number:"))
num2=20
num3=30
if num1>num2:
  print("if part run")
elif num1>num3:
    print("elif part run")
else:
  print("else part run")
```

```python
marks =int(input("Enter The Marks:"))
if marks>=90:
  print("A Grade")
elif marks>=80:
  print("B Grade")
elif marks>=70:
  print("C Grade")
elif marks>=60:
  print("D Grade")
else:
  print("FAIL")
```
