```python
class BankAccount:
    def __init__(self, password, name, balance):
        self.__password = 'devendra'
        self.name = name
        self.__balance = balance

    def check_balance(self, password):
        if password == self.__password:
            print("Current Balance =", self.__balance)
        else:
            print("Wrong Password")

    def deposit(self, password, amount):
        if password == self.__password:
            self.__balance += amount
            print(f"₹{amount} Deposit Successfully")
        else:
            print("Wrong Password")

    def withdraw(self, password, amount):
        if password == self.__password:
            if amount <= self.__balance:
                self.__balance -= amount
                print(f"₹{amount} Withdraw Successfully")
            else:
                print("Insufficient Balance")
        else:
            print("Wrong Password")


acc = BankAccount("devendraaa", "Devendra", 10000)

acc.check_balance("devendra")
acc.deposit("devendra", 5000)
acc.withdraw("devendra", 3000)
acc.check_balance("devendra")
```

```python
class BankAccount:

    def __init__(self, name, balance):
        pin = input("Enter PIN: ")

        if pin != "1234":
            print("Wrong PIN")
            return
        self.name = name
        self.__balance = balance   # private variable
        print("Login Successful ")

    def deposit(self, amount):
        self.__balance += amount
        print(f"₹{amount} Deposit Successfully")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print(f"₹{amount} Withdraw Successfully")
        else:
            print("Insufficient Balance")

    def check_balance(self):
        print("Current Balance =", self.__balance)


# Object
acc = BankAccount("Devendra", 10000)

acc.check_balance()
acc.deposit(5000)
acc.withdraw(3000)
acc.check_balance()

print(acc.__balance)
```

#overriding vs overloading

```python
#overriding
def Student(name,roll_no):
  print(name)
  print(roll_no)
Student('Devendra',123)
```

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):   # Overriding
        print("Dog barks")


d = Dog()
d.sound()
```

```python
#overloading
class Calculator:
    def add(self, a, b=0, c=0):
        print(a + b + c)

c = Calculator()

c.add(10)         # 10
c.add(10, 20)     # 30
c.add(10, 20, 30) # 60
```
